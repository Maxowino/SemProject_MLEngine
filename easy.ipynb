```python
# Confirm that the expected column exists
required_columns = ["Question", "Answer", "Category", "Difficulty"]

for column in required_columns:
    if column not in df.columns:
        raise ValueError(f"Missing required column: {column}")

# Normalize question text for comparison
def normalize_question(question):
    return " ".join(str(question).casefold().split())

existing_questions = set(
    df["Question"].dropna().map(normalize_question)
)

# Remove duplicate questions within the new list
new_df = new_df.drop_duplicates(
    subset=["Question"],
    keep="first"
).copy()

# Exclude questions already in the original dataset
new_df["_normalized_question"] = new_df["Question"].map(
    normalize_question
)

new_df = new_df[
    ~new_df["_normalized_question"].isin(existing_questions)
].copy()

new_df = new_df.drop(columns=["_normalized_question"])

print("New questions after duplicate checking:", len(new_df))
print("\nQuestions to be added:")
display(new_df[["Question", "Category", "Difficulty"]])
```

```python
# Match the original column structure
# The original dataset may have an additional Question Number column.
if "Question Number" in df.columns:
    new_df.insert(
        0,
        "Question Number",
        range(
            int(pd.to_numeric(df["Question Number"], errors="coerce").max()) + 1,
            int(pd.to_numeric(df["Question Number"], errors="coerce").max()) + 1
            + len(new_df)
        )
    )

# Keep only the columns present in the original dataset
new_df = new_df[df.columns.intersection(new_df.columns).tolist()]

# Combine original and new questions
expanded_df = pd.concat(
    [df, new_df],
    ignore_index=True
)

print("Original rows:", len(df))
print("New questions added:", len(new_df))
print("Expanded dataset rows:", len(expanded_df))

print("\nUpdated difficulty distribution:")
print(expanded_df["Difficulty"].value_counts())

display(expanded_df.tail(10))
```

```python
OUTPUT_FILE = BASE_DIR / "data" / "expanded_software_questions.csv"

# Create the data folder if it does not exist
OUTPUT_FILE.parent.mkdir(parents=True, exist_ok=True)

expanded_df.to_csv(OUTPUT_FILE, index=False)

print("Expanded dataset saved successfully!")
print("File location:", OUTPUT_FILE.resolve())
```

```python
difficulty_counts = expanded_df["Difficulty"].value_counts()

print("Difficulty counts:")
print(difficulty_counts)

print("\nDifficulty percentages:")
print((expanded_df["Difficulty"].value_counts(normalize=True) * 100).round(2))

if "Easy" in difficulty_counts.index:
    print("\nEasy questions added to the dataset.")
    print("Total Easy questions:", difficulty_counts["Easy"])
else:
    print("\nCheck that the Difficulty values use the label 'Easy'.")
```

```python
# Standardize difficulty labels for counting
expanded_df["Difficulty"] = expanded_df["Difficulty"].astype(str).str.strip()

# Find categories shared by the dataset
difficulty_counts = expanded_df["Difficulty"].value_counts()

if not {"Easy", "Medium", "Hard"}.issubset(set(difficulty_counts.index)):
    raise ValueError(
        "The dataset must contain Easy, Medium, and Hard difficulty labels."
    )

# Use the size of the smallest class
samples_per_class = int(difficulty_counts[["Easy", "Medium", "Hard"]].min())

balanced_df = pd.concat(
    [
        expanded_df[expanded_df["Difficulty"] == level].sample(
            n=samples_per_class,
            random_state=42
        )
        for level in ["Easy", "Medium", "Hard"]
    ],
    ignore_index=True
)

# Shuffle the rows
balanced_df = balanced_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

BALANCED_FILE = BASE_DIR / "data" / "balanced_experimental_dataset.csv"

balanced_df.to_csv(BALANCED_FILE, index=False)

print("Balanced dataset saved to:", BALANCED_FILE.resolve())
print("\nBalanced difficulty distribution:")
print(balanced_df["Difficulty"].value_counts())

print("\nTotal rows:", len(balanced_df))
```

```python
# VALIDATION: Load the balanced dataset to confirm it was saved correctly

import pandas as pd
from pathlib import Path

BASE_DIR = Path.cwd()
DATA_FILE = BASE_DIR / "data" / "balanced_experimental_dataset.csv"

if not DATA_FILE.exists():
    raise FileNotFoundError(
        f"Balanced dataset not found: {DATA_FILE.resolve()}"
    )

df = pd.read_csv(DATA_FILE, encoding="cp1252")

print("Dataset loaded successfully")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
# Check that the essential columns exist
required_columns = ["Question", "Answer", "Category", "Difficulty"]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

# Check missing values
print("Missing values per column:")
print(df[required_columns].isnull().sum())

# Check blank questions and answers
for column in ["Question", "Answer", "Category", "Difficulty"]:
    df[column] = df[column].astype("string").str.strip()

blank_questions = df["Question"].isna() | df["Question"].eq("")
blank_answers = df["Answer"].isna() | df["Answer"].eq("")

print("\nBlank questions:", int(blank_questions.sum()))
print("Blank answers:", int(blank_answers.sum()))

# Normalize question text for duplicate detection
def normalize_question(question):
    return " ".join(str(question).casefold().split())

normalized = df["Question"].dropna().map(normalize_question)
duplicate_count = int(normalized.duplicated().sum())

print("\nDuplicate question texts:", duplicate_count)

# Difficulty distribution
print("\nDifficulty counts:")
print(df["Difficulty"].value_counts())

print("\nDifficulty percentages:")
print(
    (df["Difficulty"].value_counts(normalize=True) * 100)
    .round(2)
    .astype(str) + "%"
)
```

```python
# Show rows with missing or blank questions/answers and duplicate questions

print("Rows with missing or blank questions/answers:")
display(df[blank_questions | blank_answers])

print("\nDuplicate question rows:")
duplicate_mask = df["Question"].notna() & df["Question"].map(
    normalize_question
).duplicated(keep=False)

display(
    df.loc[duplicate_mask, required_columns]
    .sort_values("Question")
)

# Check class balance
counts = df["Difficulty"].value_counts()

print("\nIs the dataset balanced?")
print(
    "Yes"
    if len(counts) == 3 and counts.nunique() == 1
    else "No — check the counts above."
)
```
