```python

import pandas as pd
from pathlib import Path


from pathlib import Path

BASE_DIR = Path.cwd()
DATA_FILE = BASE_DIR / "data" / "Software Questions.csv"

print("Current folder:", BASE_DIR)
print("Dataset path:", DATA_FILE)
print("Dataset exists:", DATA_FILE.exists())

# Load the dataset
df = pd.read_csv(DATA_FILE, encoding="cp1252")

print("DATASET LOADED")
print("Rows:", len(df))
print("Columns:", df.columns.tolist())

# Keep the columns needed by the app
required_columns = ["Question", "Answer", "Category", "Difficulty"]

question_bank = df[required_columns].copy()

# Remove rows missing required information
question_bank = question_bank.dropna(subset=required_columns)

# Remove leading and trailing spaces
for col in required_columns:
    question_bank[col] = question_bank[col].astype(str).str.strip()

# Remove empty values
for col in required_columns:
    question_bank = question_bank[question_bank[col] != ""]

# Remove duplicate questions
question_bank = question_bank.drop_duplicates(
    subset=["Question"]
).reset_index(drop=True)

# Display the cleaned dataset
print("\nCLEANED DATASET")
print("Total questions:", len(question_bank))

print("\nDifficulty distribution:")
print(question_bank["Difficulty"].value_counts())

print("\nCategory distribution:")
print(question_bank["Category"].value_counts())

# Save the cleaned dataset
OUTPUT_FILE = BASE_DIR / "data" / "cleaned_software_questions.csv"
question_bank.to_csv(OUTPUT_FILE, index=False)

print("\nSaved to:", OUTPUT_FILE)
```
